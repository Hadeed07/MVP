**Text Processing**

```python
import re

query_strings = []

for spine in query:
    texts = []

    for text, score in zip(spine['text'], spine['scores']):
        # Replaces Punctuations with white spaces
        text = re.sub(r"[^\w\s]", " ", text)

        # Split CamelCase
        text = re.sub(r"([a-z])([A-Z])", r"\1 \2", text)

        # Collapse whitespace
        text = re.sub(r"\s+", " ", text)

        # Trim
        text = text.strip()

        # Lowercase
        text = text.lower()

        # Remove garbage tokens: drop any single-character token,
        # ASCII or not (a lone stray character is noise either way)
        filtered = [token for token in text.split() if len(token) > 1]

        if filtered:
            texts.append(text)

    query_strings.append(" ".join(texts))

query_strings
```

**Text Processing with WordNinja**

```python
import re
import wordninja

query_strings = []

for spine in query:
    texts = []

    for text, score in zip(spine["text"], spine["scores"]):

        # Optional confidence filter
        if score < 0.4:
            continue

        # Replace punctuation with spaces
        text = re.sub(r"[^\w\s]", " ", text)

        # Split CamelCase
        text = re.sub(r"([a-z])([A-Z])", r"\1 \2", text)

        # WordNinja
        words = []
        for token in text.split():
            if " " not in token and len(token) >= 8:
                words.extend(wordninja.split(token))
            else:
                words.append(token)

        text = " ".join(words)

        # Remove single-character garbage
        filtered = [token for token in text.split() if len(token) > 1]
        text = " ".join(filtered)

        # Normalize whitespace
        text = re.sub(r"\s+", " ", text).strip()

        # Lowercase
        text = text.lower()

        if text:
            texts.append(text)

    query_strings.append(" ".join(texts))

query_strings
```

**WordNinja Examples**

```python
import wordninja

# 1. Mixed Capitalization (It ignores casing and outputs lowercase)
print(wordninja.split('DataScienceAndMachineLearning'))
# Output: ['data', 'science', 'and', 'machine', 'learning']

# 2. Words Containing Numbers (Numbers are split out cleanly)
print(wordninja.split('covid19response2026'))
# Output: ['covid', '19', 'response', '2026']

# 3. Overlapping Ambiguous Words (Chooses the most probable English words)
print(wordninja.split('therapist')) 
# Output: ['the', 'rapist'] <-- Note: A classic NLP challenge. It prefers shorter, high-frequency words.

print(wordninja.split('psychotherapist'))
# Output: ['psycho', 'therapist'] <-- Context changes the probability!
```

```python
import wordninja

# Split simple jammed text
words1 = wordninja.split('derekanderson')
print(words1)  # Output: ['derek', 'anderson']

# Split a sentence without spaces
words2 = wordninja.split('imateapot')
print(words2)  # Output: ['im', 'a', 'teapot']

# Join the split words back into a readable sentence
sentence = " ".join(wordninja.split('thequickbrownfoxjumpsoverthelazydog'))
print(sentence)  # Output: 'the quick brown fox jumps over the lazy dog'
```
